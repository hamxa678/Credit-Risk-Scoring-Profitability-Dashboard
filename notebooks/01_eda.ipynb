# 01 — Exploratory Data Analysis
Load the Home Credit application table and inspect target imbalance, missingness, and core financial variables.

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('../data/raw/application_train.csv')
display(df.head())
display(df['TARGET'].value_counts(normalize=True).rename('share'))
```

```python
missing = df.isna().mean().sort_values(ascending=False).head(20)
display(missing.to_frame('missing_rate'))
```
