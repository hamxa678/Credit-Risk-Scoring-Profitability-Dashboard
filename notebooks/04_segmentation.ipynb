# 04 — Customer Segmentation
Create compact customer-level features and cluster customers with K-Means.

```python
import pandas as pd
from src.features.build_features import add_domain_features, make_customer_features
from src.segmentation.clustering import fit_segments
df = pd.read_csv('../data/raw/application_train.csv')
df = add_domain_features(df)
features = make_customer_features(df)
model, labels = fit_segments(features, n_clusters=4)
features['segment'] = labels
features.groupby('segment').mean(numeric_only=True)
```
